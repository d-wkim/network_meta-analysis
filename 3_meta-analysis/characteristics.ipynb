Each **dataset** has:
- long-format (extracted), as extracted and transformed
- wide-format (reformatted), with one row per study
- sheets for analysis to be used in R for analyses
- publication-ready tables for copy-pasting into manuscripts

**Study characteristics**

```python
import pandas as pd

char = pd.read_excel(f"./data/tables/xlsx/characteristics.xlsx")
char.to_csv(f"./data/extracted/csv/characteristics.csv")
df = char

df["group"] = df.groupby(["id"]).cumcount() + 1
    
wide = (
    df.pivot(
        index= "id",
        columns="group",
        values=["study", "subgroup", "total", "age", "male", "female", "follow_up"]
    )
).reset_index()

wide.columns = [
    f"{name}{group}" if name != "subgroup" else f"treat{group}"
    for name, group in wide.columns
]
    
char = wide
char = char.drop(columns = "study2").fillna("")
char = (
    char.rename(columns={
        "treat1": "subgroup1",
        "treat2": "subgroup2",
        "total1": "n1",
        "total2": "n2",
    })
)

char.to_csv(f"./data/reformatted/csv/characteristics.csv")
char.head()
```
