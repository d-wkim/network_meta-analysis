**`p_sd`**: *python function for converting sample size, `n` and number of events, `k` into standard deviation. Returns proportions, standard error, and standard deviation.*

```python
n = 16
k = 4

def p_sd(k, n):
    import numpy as np
    p = (k + 0.5)/(n + 0.5)
    se = p/(1-p)
    sd = se * np.sqrt(n)
    p, se, sd = [round(x, 3) for x in (p, se, sd)]
    print(f"""\
p  = {p}
se = {se}
sd = {sd} \
""")
    return sd
```

```python
sd = p_sd(4, 16)
```

```python
from scipy.stats import norm

def range_sd(n, a, b):
    range = b - a
    z = ((n-0.375)/(n+0.25))
    sd = 2 * norm.ppf(z)
    sd = round(sd, 3)
    print(sd)
    return sd
```

```python
def hozo(n, a, b):
    range = b - a
    if n <= 15:
        sd = range/4
    if n >= 15 and n < 70:
        sd = range/6
    sd = round(sd, 3)
    print(sd)
    return sd
```
